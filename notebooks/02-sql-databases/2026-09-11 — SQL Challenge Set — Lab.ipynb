### Q1 — Every track with its artist's name and country.

*Expected: 9 rows, one per track.*

```python
q1 = q('''SELECT
    tracks.title,
    artists.name AS artist,
    artists.country
FROM tracks
JOIN artists
    ON tracks.artist_id = artists.artist_id;''')
q1
```

For this question I joined the tracks and artist tables using artist_id so that I could show all of the tracks and their artists and country.

### Q2 — Which genre has the longest average track length?

Return the genre and the average, not just the name.

```python
q(''' SELECT
    genre,
    AVG(seconds) AS avg_seconds
FROM tracks
WHERE genre IS NOT NULL
GROUP BY genre
ORDER BY avg_seconds DESC
LIMIT 1;''')
```

For this question, I first grouped the tracks by their genres and then I calculated the average track length. After that, I sorted the averages from highest to lowest so I could be able to spot the genre with the longest average length.

### Q3 — For each user: how many plays, and how many distinct tracks?

Someone who played one track four times is a different listener from someone who played four different tracks. Your result should make that visible.

```python
q3 = q(''' SELECT
    user,
    COUNT(*) AS plays,
    COUNT(DISTINCT track_id) AS distinct_tracks
FROM plays
GROUP BY user
ORDER BY user;''')
q3
```

For this question, I grouped the plays by user and I counted the total plays and distinct tracks ids so I could see how many times each user listend for and how many different tracks they listened to.

### Q4 — Which tracks have never been played?

*Expected: 2 rows.* Hint: `LEFT JOIN` and then keep the rows where the right side came back `NULL`.

```python
q4 = q(''' SELECT
    tracks.track_id,
    tracks.title
FROM tracks
LEFT JOIN plays
    ON tracks.track_id = plays.track_id
WHERE plays.track_id IS NULL;''')
q4
```

For this question, I used left join between track and plays and looked for and Null matches to determine any tracks that have not been played.

### Q5 — Rank artists by total listening time.

Sum the seconds actually listened across all plays, most to least, and include a minutes column rounded to one decimal.

```python
q(''' SELECT
    artists.name AS artist,
    SUM(tracks.seconds) AS total_seconds,
    ROUND(SUM(tracks.seconds) / 60.0, 1) AS total_minutes
FROM plays
JOIN tracks
    ON plays.track_id = tracks.track_id
JOIN artists
    ON tracks.artist_id = artists.artist_id
GROUP BY artists.artist_id, artists.name
ORDER BY total_seconds DESC;''')
```

For this question, I joined plays, tracks, and artists and then I determined the sum of the track seconds for each of the artists. Then I used this information to rank the artists by the total amount of time that users spent listening to them.

### Q6 — Which tracks are missing a genre?

Return the track id and title. Then, in a comment, say what `WHERE genre != 'Pop'` would have done to these rows and why.

```python
q('''SELECT
    track_id,
    title
FROM tracks
WHERE genre IS NULL;''')
```

For this question, I used the "WHERE genre is NULL" to find any tracks that had missing genre information as that is how SQL checks for missing values.

### Q7 — Plays per day.

`played_on` is stored as text like `'2026-09-01'`. Count plays per date, earliest first, and include the number of distinct users active that day.

```python
q('''SELECT
    played_on,
    COUNT(*) AS play_count,
    COUNT(DISTINCT user) AS distinct_users
FROM plays
GROUP BY played_on
ORDER BY played_on ASC;''')
```

For this question, I grouped the plays by date and then I counted total plays and distinct users so I could see how much users were listening to music for everyday and by how many different users.
